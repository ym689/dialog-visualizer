## No Credibility

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
import os
import matplotlib

# 数据
labels = ['inspired', 'redial']
categories = ['with credibility', 'without credibility']
credibility = [[3.859, 2.819], [1, 1]]

x = np.arange(len(labels))  # x轴位置
width = 0.35  # 柱宽

fig, ax = plt.subplots(figsize=(6, 4))

# 绘制柱状图（常规学术风格）
rects1 = ax.bar(x - width/2, [c[0] for c in credibility], width, label='with credibility', color='#bfdeac')
rects2 = ax.bar(x + width/2, [c[1] for c in credibility], width, label='without credibility', color='#e2f0d9')

# 添加标签
ax.set_ylabel('Credibility', fontsize=16)
ax.set_xticks(x)
ax.set_xticklabels(labels, fontsize=14)
ax.set_ylim(0, 5)

# 在柱子上写数值
for rect in rects1 + rects2:
    height = rect.get_height()
    ax.annotate(f'{height:.2f}',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3),  # 3 points vertical offset
                textcoords="offset points",
                ha='center', va='bottom', fontsize=12)

# 保留图例，调整字体大小
ax.legend(fontsize=12)

# 美化边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig("/home/yanming/dialog-visualizer/imgs/nocre.pdf",bbox_inches='tight')
plt.show()
```
